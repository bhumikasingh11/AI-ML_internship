```python
#operators
#arithmetic:
a,b=2,4
print(a,b)
print("Addition:",a+b)
print("Multiplication",a*b)
print("division:",a/b)
print("modulus:",a%b)
print("exponent:",a**b)
#comparisons:
x=10
y=20
print(x,y)
print("equal?",x==y)
print("not equal:",x!=y)
print("greater than:",x>y)
print("less than:",x<y)
print(f"Greater or equal (>=): {x >= y}")
print(f"Less or equal (<=): {x <= y}")

#logical operator
is_raining = True
has_umbrella = False
is_weekend = True

print(f"\nLogical operators:")
print(f"AND: {is_raining and has_umbrella}")  # Both must be True
print(f"OR: {is_raining or is_weekend}")  # At least one True
print(f"NOT: {not is_raining}")  # Reverse the value

#assignment operator:
value=10
value=value+5
value=value-3
print(value)
value=value/4
print(value)

#strings
message="python is cool"
print(message.lower())
print(message.upper())
print(message.capitalize())
print(message.title())
print("length:",len(message))
print(message.replace("cool","awesome"))
print(message.find("awesome"))
```

```python
#string concatination and repetition
first_name = "John"
last_name = "Smith"
full_name = first_name + " " + last_name
print(f"\nConcatenation: {full_name}")
print(f"Repetition: {'Ha' * 3}")
#lists
#print("lists"+"--->"+"ordered"+"mutable"+"duplicate allowed+"can store different data types"+"indexed"+"dynamic size"+)
fruits=["apple","banana","orange","mango"]
numbers=[1,2,3,4,5]
mixed= [1, "hello", 3.14, True]
print(f"Fruits list: {fruits}")
print(f"Numbers list: {numbers}")
print(f"Mixed list: {mixed}")

# Indexing (accessing elements)
print(f"\nFirst fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")
print(f"Second fruit: {fruits[1]}")

# Slicing lists
print(f"\nFirst 2 fruits: {fruits[0:2]}")
print(f"Fruits from index 1 to 3: {fruits[1:4]}")
print(f"Last 2 fruits: {fruits[-2:]}")
print(f"Every other fruit: {fruits[::2]}")

# List Methods
print(f"\nOriginal fruits: {fruits}")
fruits.append("grape")
print(f"After append('grape'): {fruits}")
fruits.insert(1, "kiwi")
print(f"After insert(1, 'kiwi'): {fruits}")
fruits.remove("banana")
print(f"After remove('banana'): {fruits}")
popped = fruits.pop()
print(f"After pop(): {fruits}, Removed: {popped}")
fruits.sort()
print(f"After sort(): {fruits}")
fruits.reverse()
print(f"After reverse(): {fruits}")

```

```python
#dictionaries:
# Creating dictionaries
student = {
    "name": "Alice Johnson",
    "age": 20,
    "course": "Computer Science",
    "gpa": 3.8,
    "is_active": True
}

print(f"Student dictionary: {student}")
print(f"Student name: {student['name']}")
print(f"Student age: {student['age']}")

# Dictionary methods
print(f"\nAll keys: {student.keys()}")
print(f"All values: {student.values()}")
print(f"All items: {student.items()}")

# Adding/updating values
student["email"] = "alice@university.com"
print(f"\nAfter adding email: {student}")
student["gpa"] = 3.9
print(f"After updating GPA: {student}")

# Safe access with get()
print(f"Phone (default): {student.get('phone', 'Not provided')}")

# Looping through dictionary
print("\nIterating through dictionary:")
for key, value in student.items():
    print(f"  {key}: {value}")

```

```python
#palindrome:
#approach:adding the new character at the front of string by travrsing
s1=input("enter a word:")
rev=""
for ch in s1:
  rev=ch+rev
if s1==rev:
    print("palindrome")
else:
    print("not palindrome")
    #approach2
word = input("Enter a word: ")

left = 0
right = len(word) - 1

is_palindrome = True

while left < right:
    if word[left] != word[right]:
        is_palindrome = False
        break
    left += 1
    right -= 1

if is_palindrome:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#count vowels in a string
s1=input("enter a word:")
vowels=["a","e","i","o","u"]
s1=s1.lower()
count=0
for ch in s1:
  if ch in vowels:
    count=count+1
print("number of vowels are:",count)

```
